#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
# Select the variables I needed and remove na values
# seed, reg_wins_pct, total_wins_pct, and tourney_wins are numeric variables.
# bid is a categorical variable.

tidy_data_na <- historical_record |>
    select(seed, reg_wins_pct, total_wins_pct, bid, tourney_wins)

tidy_data <- tidy_data_na |>
    filter(!is.na(seed), 
           !is.na(reg_wins_pct), 
           !is.na(total_wins_pct), 
           !is.na(bid),
           !is.na(tourney_wins))

# Show the first 6 rows
head(tidy_data)
```

```r
# Plot the distribution of tourney_wins using bar graph
tourney_wins_graph <- tidy_data |>
  ggplot(aes(x = factor(tourney_wins))) +
  geom_bar(fill = "steelblue") +
  labs(title = "Distribution of Tournament Wins",
       x = "Number of Tournament Wins",
       y = "Number of Schools")

# Show the output
tourney_wins_graph
```

```r
# Count the number of schools in each bid category
bid_counts <- tidy_data |>
  group_by(bid) |>
  summarize(count = n())

# Calculate the proportion for each bid category
bid_proportions <- bid_counts |>
  mutate(proportion = count / sum(count))

# Plot the relative proportion of each bid category
bid_relative_graph <- 
  ggplot(bid_proportions, aes(x = bid, y = proportion)) +
  geom_col(fill = "darkorange") +
  labs(
    title = "Proportion of Schools by Bid Type",
    x = "Bid Type",
    y = "Proportion of Schools")

# Show the output
bid_relative_graph
```
